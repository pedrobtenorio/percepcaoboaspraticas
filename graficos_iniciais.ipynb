# Graficos iniciais (good vs bad)

Este notebook usa o CSV numerico gerado em `transformacao_dados.ipynb`.


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

## Carregar dataset numerico


```python
csv_path = Path("respostas_form_NUMERIC.csv")
try:
    df = pd.read_csv(csv_path, encoding="utf-8")
except UnicodeDecodeError:
    df = pd.read_csv(csv_path, encoding="latin-1")
df.head()
```

```python
# Verificar nomes das colunas apos leitura
list(df.columns)
```

```python
# Normaliza espacos extras nos nomes das colunas
df.columns = df.columns.str.strip()
df.head()
```

## Organizar em formato long


```python
likert_cols = [c for c in df.columns if c.endswith(("_FE", "_NC", "_EE"))]
long = df.melt(value_vars=likert_cols, var_name="item_dim", value_name="score")
long["dim"] = long["item_dim"].str[-2:]
long["item"] = long["item_dim"].str[:-3]
long.head()
```

## Pares good vs bad
Estamos emparelhando codigos que seguem as boas praticas e codigos que não as seguem, o objetivo é comparar as notas medias entre esses pares.


```python
pairs = [
    {"label": "OCP (shape)", "good": "shape_ocp_good", "bad": "ocp_bad_shape"},
    {"label": "SRP (geral)", "good": "srp_good", "bad": "SRP_bad"},
    {"label": "KISS (sum)", "good": "sum_e_kiss_g", "bad": "SumEvenNumbers _kiss_bad"},
    {"label": "YAGNI (payment)", "good": "payment_YAGNI_g", "bad": "YAGNI_bad"},
    {"label": "Modular (calculator)", "good": "MODULAR_cal_good", "bad": "Calculator_modu_ruim"},
    {"label": "Nomes (celsius)", "good": "celcius_good_names", "bad": "celcius_bad_names"},
    {"label": "DRY", "good": "dry_good", "bad": "DRY_bad"},
    {"label": "Nomes (Palindromo) ", "good": "palindromo_clean", "bad": "var_name_bad"},
]
```

## Calcular diferenca media (good - bad) por dimensao


```python
rows = []
for pair in pairs:
    for dim in ["FE", "NC", "EE"]:
        good_scores = long[(long["item"] == pair["good"]) & (long["dim"] == dim)]["score"]
        bad_scores = long[(long["item"] == pair["bad"]) & (long["dim"] == dim)]["score"]
        rows.append({
            "label": pair["label"],
            "dim": dim,
            "good_mean": good_scores.mean(),
            "bad_mean": bad_scores.mean(),
            "delta": good_scores.mean() - bad_scores.mean(),
        })
pair_df = pd.DataFrame(rows)
pair_df
```

## Grafico: diferenca media por dimensao


```python
dim_labels = {
    "FE": "Facilidade de entendimento",
    "NC": "Confianca para modificar",
    "EE": "Elegancia e eficiencia",
}

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
for i, dim in enumerate(["FE", "NC", "EE"]):
    ax = axes[i]
    data = pair_df[pair_df["dim"] == dim]
    ax.barh(data["label"], data["delta"], color="#4C78A8")
    ax.axvline(0, color="#333333", linewidth=1)
    ax.set_title(dim_labels[dim])
    ax.set_xlabel("Diferenca media entre boas e mas praticas")

fig.tight_layout()
plt.show()
```

# Pivot table por perfil (RQ2)

Exemplo de cruzamento entre perfis e percepcao (medias).


## Selecionar colunas de perfil


```python
profile_cols = [
    "  Como você avalia o seu nível de experiência em programação?  ",
    "  Qual é o seu nível de familiaridade com boas práticas de programação (por exemplo, clean code, padrões de projeto)?  ",
    "  Qual sua situação Profissional/Acadêmica  ",
    "  Com que frequência você trabalha em equipe em projetos de programação?  ",
    "  Qual é o seu nível de escolaridade na área de TI  ",
    "  Com que frequência você participa de revisões de código (code reviews)  ",
]

[c for c in profile_cols if c in df.columns]
```

## Tabela long com perfis


```python
long_profiles = df.melt(
    id_vars=profile_cols,
    value_vars=likert_cols,
    var_name="item_dim",
    value_name="score",
)
long_profiles["dim"] = long_profiles["item_dim"].str[-2:]
long_profiles["item"] = long_profiles["item_dim"].str[:-3]
long_profiles.head()
```

## Pivot: media por perfil e dimensao

Troque `perfil` abaixo pela coluna que voce quer analisar primeiro.


```python
dim_labels = {
    "FE": "Facilidade de entendimento",
    "NC": "Confianca para modificar",
    "EE": "Elegancia e eficiencia",
}

# Pivot table para cada perfil
pivots = {}
for perfil in profile_cols:
    table = pd.pivot_table(
        long_profiles,
        index=perfil,
        columns="dim",
        values="score",
        aggfunc="mean",
    )
    table = table.rename(columns=dim_labels)
    pivots[perfil] = table

# Exibir todos (ou escolha um especifico)
pivots
```

```python
from IPython.display import display

for perfil, table in pivots.items():
    print(perfil.strip())
    styled = table.style.format(precision=3).set_caption("Media por dimensao")
    display(styled)
    print()
```
