# Transformacao de dados (Likert)

Notebook focado apenas na conversao das respostas textuais para escala numerica.


## Mapeamentos Likert

Mapeamentos acordados para FE (facilidade), NC (confianca) e EE (elegancia/eficiencia).


```python
import pandas as pd
from pathlib import Path

# FE (Facilidade)
LIKERT_FE_ORDER = [
    ("Muito Difícil – Exige grande esforço e tempo para deduzir o que o código faz", -2),
    ("Difícil – Demanda esforço considerável para compreender a funcionalidade", -1),
    ("Neutro", 0),
    ("Fácil – O propósito e o funcionamento são claros e rapidamente identificáveis", 1),
    ("Muito Fácil – O código é autoexplicativo, o propósito salta aos olhos", 2),
]

# NC (Confianca)
LIKERT_NC_ORDER = [
    ("Nenhuma Confiança – Praticamente certo de que qualquer mudança introduzirá bugs", -2),
    ("Pouca Confiança – Há um alto risco de causar problemas ao modificar", -1),
    ("Neutro", 0),
    ("Confiante – Sinto-me seguro para fazer alterações com testes razoáveis", 1),
    ("Muito Confiante – Consigo alterar o código com tranquilidade, minimizando riscos", 2),
]

# EE (Elegancia/Eficiencia)
LIKERT_EE_ORDER = [
    ("Nenhuma Elegância/Eficiência – A solução é confusa, prolixa ou ineficiente", -2),
    ("Pouca Elegância/Eficiência – A solução tem falhas notáveis em clareza, concisão ou uso de recursos", -1),
    ("Neutro", 0),
    ("Boa Elegância/Eficiência – A solução é bem pensada, clara e faz bom uso dos recursos", 1),
    ("Excelente Elegância/Eficiência – A solução é um modelo de clareza, concisão e desempenho", 2),
]

LIKERT_FE_MAP = dict(LIKERT_FE_ORDER)
LIKERT_NC_MAP = dict(LIKERT_NC_ORDER)
LIKERT_EE_MAP = dict(LIKERT_EE_ORDER)

LIKERT_MAPS = {
    "FE": LIKERT_FE_MAP,
    "NC": LIKERT_NC_MAP,
    "EE": LIKERT_EE_MAP,
}
```

## Carregar CSV

Ajuste o caminho se necessario.


```python
csv_path = Path("respostas_form_CLEANED.csv")
df = pd.read_csv(csv_path)
df.head()
```

## Transformar colunas Likert

Converte colunas com sufixo `_FE`, `_NC`, `_EE` para escala numerica.


```python
def map_likert_column(series: pd.Series) -> pd.Series:
    suffix = series.name.split("_")[-1]
    mapping = LIKERT_MAPS.get(suffix)
    if mapping is None:
        return series
    return series.map(mapping)

likert_cols = [c for c in df.columns if c.endswith(("_FE", "_NC", "_EE"))]
df_numeric = df.copy()
df_numeric[likert_cols] = df_numeric[likert_cols].apply(map_likert_column)

df_numeric[likert_cols].head()
```

## Checagem rapida

Verifique se existem valores nao mapeados (NaN) apos a conversao.


```python
unmapped_counts = df_numeric[likert_cols].isna().sum().sort_values(ascending=False)
unmapped_counts.head(10)
```

## Salvar dataset numerico

Gera um novo CSV com as colunas Likert convertidas para numeros.


```python
output_path = Path("respostas_form_NUMERIC.csv")
df_numeric.to_csv(output_path, index=False)
output_path
```
